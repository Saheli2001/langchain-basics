```python
## In case of source based retriever the fetching is done by keyword matching usually. Not semantic matching. In case of vector store retriver we search using semantic similarity
## between query embedding and doc embedding
```

```python
## Vector Store Retriever

##1. Store documents into a vector store like FAISS, CHROMA

##2. Each document is converted into a dense vector using an Embedding Model.

## when a query is made, 

## query is tored into a vector

## retriever compares query vector with the stored vector

##comares the similarity between the query vector and the document vectors using a similarity metric (like cosine similarity).

## Retrievs topk similary results
```

```python
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
documents = [
    Document(page_content="This is the content of document 1.", metadata={"source": "doc1"}),
    Document(page_content="This is the content of document 2.", metadata={"source": "doc2"}),
]
```

```python
embeddings = OpenAIEmbeddings()

vectorstore = Chroma.from_documents(documents, embeddings, collection_name="my_collection")
```

```python
retriever = vectorstore.as_retriever(search_type="similarity", search_kwargs={"k": 1})
```

```python
query = "What is LangChain?"
results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"Result {i+1}:")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}")
    print()
```

```python
vectorstore.similarity_search(query, k=1) # not works for different search strategy, retriever gives  more flexibility to change search strategy and other parameters
```

```python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
documents = [
    Document(page_content="The Grand Canyon was carved over millions of years by the Colorado River.", metadata={"source": "doc1"}),
    Document(page_content="Photosynthesis allows green plants to convert light energy into chemical energy.", metadata={"source": "doc2"}),
    Document(page_content="The Grand Canyon is located in northern Arizona and is known for its layered rock formations.", metadata={"source": "doc3"}),
    Document(page_content="Plants use sunlight, carbon dioxide, and water to produce glucose during photosynthesis.", metadata={"source": "doc4"}),
    Document(page_content="The Colorado River continues to flow through the Grand Canyon today.", metadata={"source": "doc5"}),
    Document(page_content="Chlorophyll in plant cells absorbs light energy needed for photosynthesis.", metadata={"source": "doc6"}),
    Document(page_content="The Grand Canyon provides habitats for many different species of plants and animals.", metadata={"source": "doc7"}),
    Document(page_content="Oxygen is released as a byproduct when plants carry out photosynthesis.", metadata={"source": "doc8"}),
]
```

```python
embedding_model = OpenAIEmbeddings()
vectorstore = FAISS.from_documents(documents, embedding_model)
base_retriever = vectorstore.as_retriever(search_kwargs={"k": 2})
llm = ChatOpenAI(model_name="gpt-3.5-turbo", temperature=0)
compressor = LLMChainExtractor.from_llm(llm=llm)
query = "what is photosynthesis?"
compressed_retriever = ContextualCompressionRetriever(base_compressor=compressor, base_retriever=base_retriever)
compressed_results = compressed_retriever.invoke(query)
```
